# 📊 Population Age and Gender Distribution Analysis

This project analyzes population distribution across different age groups and genders using Python (Pandas & Matplotlib).

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt


# Sample dataset
data = {
    'Age Group': ['0-10','11-20','21-30','31-40','41-50','51-60','61+'],
    'Male': [50,60,80,70,55,40,30],
    'Female': [45,55,75,68,60,42,35]
}

df = pd.DataFrame(data)
df.set_index('Age Group', inplace=True)

# Display data
print(df)
```

```python
# Plot bar chart
df.plot(kind='bar')
plt.title('Population Distribution by Age and Gender')
plt.xlabel('Age Group')
plt.ylabel('Population')
plt.legend()
plt.show()
```
